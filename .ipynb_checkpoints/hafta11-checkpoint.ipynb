# Shapiro-Wilk Testi

### Ne İşe Yarar?
Verilerinizin normal dağılıma uyup uymadığını test eder.
### Neden Önemli?
Birçok istatistiksel test (t-test, ANOVA, regresyon) normallik varsayımı gerektirir. Shapiro-Wilk bu varsayımı kontrol eder.


*bu test, istatistiksel analiz yol haritanızda size "sağa mı git (parametrik), sola mı git (non-parametrik)" diye yol gösteren bir trafik polisi gibidir.*

### örnek kod denemesi *(hoca herkesin kod denemesini istedi.)*

````python
# bu kod hocanın notlarında var. 

import numpy as np 
from scipy import stats 

# senaryo: küçük bir veri setinin normalliğini test et (n=30)
np.random.seed(15)
data = np.random.normal(loc=10,scale=2,size=30) # normal dağılımdan üretitilr. 

# shapiro wild testi uygulaması 
shapiro_statistics , p_value = stats.shapiro(data)

print(f"shapiro wild istatistiği: {shapiro_statistics:.3f}")
print(f"p değeri: {p_value:.5f}")

#hipotez karar mekanizması (alfa = 0.05)
if p_value < 0.05:
    print("\nsonuç: p değeri < 0.05\nH_0 reddilir.")
else:
    print("\nsonuç: p değeri > 0.05\nH_1 reddedilemez.")

````

### Adım Adım Açıklama:
* Veri Üretme: İlk olarak, normal dağılıma sahip olduğu bilinen 30 tane rastgele sayı üretiliyor. Bu, testin doğru çalışıp çalışmadığını görmek için yapılıyor.
* Shapiro-Wilk Testi: stats.shapiro(data) komutu ile bu 30 sayının normal dağılıma uyup uymadığı test ediliyor.
* Sonuçları Yazdırma: Testin sonucunda iki değer elde ediliyor:
    - İstatistik Değeri: Verinin normal dağılıma ne kadar benzediğini gösteren bir sayı.
    - p-değeri: Testin asıl önemli sonucu.


# Kolmogorov-Smirnov (K-S) Testi

Ne İçin Kullanılır?
Verilerin teorik bir dağılıma uygunluğunu test eder. Genellikle büyük örneklem büyüklüklerinde (n≥50) normalliği test etmek için kullanılır.

* K-S Testi, tıpkı Shapiro-Wilk testi gibi, bir veri setinin belirli bir teorik dağılıma (genellikle normal dağılıma) uyup uymadığını kontrol eden bir "uyum iyiliği" testidir.
Temel Soru: "Elimdeki verinin dağılımı, beklediğim teorik dağılımdan (örneğin normal dağılımdan) anlamlı derecede farklı mı?"


### Shapiro-Wilk vs. Kolmogorov-Smirnov Karşılaştırma Tablosu

| Özellik | Shapiro-Wilk Testi | Kolmogorov-Smirnov Testi |
| :--- | :--- | :--- |
| **Ana Amaç** | Bir verinin **normal dağılıma** uyup uymadığını test eder. | Bir verinin **herhangi bir teorik dağılıma** (normal, uniform, vb.) uyup uymadığını test eder. |
| **Uzmanlık Alanı** | Özellikle **normallik testi** için tasarlanmıştır. | Daha genel amaçlı bir "uyum iyiliği" testidir. |
| **Güç / Hassasiyet** | Normallik testinde genellikle **daha güçlü ve hassastır**. | Normallik testinde genellikle **daha az güçlüdür**. |
| **İdeal Örneklem Boyutu** | Genellikle **küçük örneklemler (n < 50)** için daha güvenilirdir. | Genellikle **büyük örneklemler (n > 50)** için tercih edilir. |
| **Yorumlama (p-değeri)** | **p > 0.05:** Normal dağılıyor.<br>**p < 0.05:** Normal dağılmıyor. | **p > 0.05:** Normal dağılıyor.<br>**p < 0.05:** Normal dağılmıyor. |
| **Pratik Kullanım** | Analizden önce **sadece normallik varsayımını** kontrol etmek için en yaygın ve güvenilir tercihtir. | İki farklı veri setinin dağılımlarını karşılaştırmak veya normallik dışındaki dağılımları test etmek için daha kullanışlıdır. |

### Özetle

*   **Sadece normallik testi yapacaksanız:** Genellikle **Shapiro-Wilk** daha iyi bir seçimdir.
*   **İki grubun dağılımını karşılaştıracaksanız veya normal dağılım dışında bir dağılımı test edecekseniz:** **Kolmogorov-Smirnov** daha esnektir.

shapiro wild: 
* Bir verinin normal dağılıma uyup uymadığını test eder
* küçük örnkeler

# Anderson-Darling (A-D) Testi

* geniş kuyruklu.
* eksterm olayları inceliyoresan (hassas veriler varsa) örn. finansl veriler

### Anderson-Darling Testi: K-S Testi'nin geliştirilmiş bir versiyonudur. Dağılımın kuyruklarına (uç noktalarına) daha fazla ağırlık vererek sapmalara karşı daha hassas sonuçlar üretir.


* Kuyruk Hassasiyeti İhtiyacı: Dağılımın uç noktalarındaki (kuyruklarındaki) aykırı değerlerin veya sapmaların normallik test sonucunu etkilemesi istendiğinde kullanılır. Bu, finansal veriler, mühendislik toleransları veya ekstrem olayların incelendiği alanlar için önemlidir.
* Alternatif Dağılımları Test Etme: A-D testi, sadece Normal dağılımı değil, aynı zamanda Log-Normal, Weibull, Gamma gibi diğer belirli dağılımların uygunluğunu test etmek için de yaygın olarak kullanılır.
* Shapiro-Wilk Alternatifi: Genellikle Shapiro-Wilk testi, küçük örneklemler için en güçlü test kabul edilirken, A-D testi de hem küçük hem de orta büyüklükteki örneklemler için oldukça güçlü bir alternatiftir.

**Kolmogorov-Smirnov Testi ise genellikle büyük örneklemler (n ≥ 50) için daha çok tercih edilir.**
Kolmogorov-Smirnov Testi ise genellikle büyük örneklemler (n ≥ 50) için daha çok tercih edilir.

# Q-Q Grafiği (Quantile-Quantile Plot)

### Q-Q Grafiği (Quantile-Quantile Plot), iki olasılık dağılımının benzerliğini görsel olarak değerlendirmek için kullanılan bir istatistiksel grafiktir. Genellikle bir veri setinin belirli bir teorik dağılıma (örneğin, normal dağılıma) uyup uymadığını kontrol etmek amacıyla kullanılır. Bu bağlamda en yaygın kullanımı normal Q-Q grafiğidir.

Nasıl Çalışır?
1. Y eksenine elinizdeki verinin değerlerini (kantillerini) koyar.
2. X eksenine ise, eğer veri tam olarak normal dağılsaydı alması gereken teorik değerleri (kantilleri) koyar.

<ul>
    <li><b>Normallik Varsayımını Kontrol Etme:</b> Bir istatistiksel testin (örneğin T-Testi) ön koşulu olarak verinin Normal dağılıma uygun olup olmadığını görmek için.
    </li>
    <li><b>Dağılımın Şeklini Anlama:</b> İstatistiksel testlerin (A-D, Shapiro-Wilk) sadece bir p-değeri verdiği durumda, Q-Q grafiği size dağılımın neden Normal olmadığını (çarpık mı, basık mı, kuyrukları mı ağır?) gösterir.
    </li>
    <li><b>Aykırı Değerleri Tespit Etme:</b> Grafiğin uç noktalarından (kuyruklarından) çok uzaklaşan noktalar, veri setindeki aykırı değerlere (outliers) işaret edebilir.
    </li>
</ul>

**Grafiğin Yorumlanması**</br>
<ul>
    <li><b>Mükemmel Uyum:</b> Veri noktalarının tamamı, grafikteki 45 derecelik düz referans çizgisine yakın bir şekilde diziliyorsa, veri seti Normal dağılım göstermektedir.
    </li>
    <li><b>Kuyruk Sapması (Basıklık/Kurtosis):</b> Noktalar düz çizgiyi takip eder, ancak uçlarda (kuyruklarda) çizgiden saparak yukarı veya aşağı kıvrılırsa, dağılımın ağır kuyruklu (leptokurtik) veya hafif kuyruklu (platykurtik) olduğu anlaşılır.
    </li>
    <li><b>Eğrilik (Çarpıklık/Skewness):</b> Noktalar S-şeklinde bir eğri oluşturuyorsa, bu verinin çarpık (skewed) olduğunu gösterir.
    </li>
</ul>

````python
# hocanın notundan aldığım bir örnek: 
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import pandas as pd

# --- Veri Setleri ---
np.random.seed(42)

# 1. Normal Dağılan Veri (Q-Q grafiğinde düz çizgi beklenir)
veri_normal = np.random.normal(loc=100, scale=10, size=100) 

# 2. Çarpık Dağılan Veri (Q-Q grafiğinde eğrilik beklenir)
veri_carpik = np.random.exponential(scale=5, size=100)

### A. Normal Dağılan Veri İçin Q-Q Grafiği

plt.figure(figsize=(12, 5))

# Sol Grafiği (Normal Veri)
plt.subplot(1, 2, 1)
stats.probplot(veri_normal, dist="norm", plot=plt)
plt.title("Normal Dağılan Veri (Beklenen: Düz Çizgi)")
plt.xlabel("Teorik Kantiller")
plt.ylabel("Örneklem Kantilleri")
 # Bu kısım görsel olarak düz bir çizgiye benzeyecektir.

# Sağ Grafiği (Çarpık Veri)
plt.subplot(1, 2, 2)
stats.probplot(veri_carpik, dist="norm", plot=plt)
plt.title("Çarpık Dağılan Veri (Beklenen: Eğri Çizgi)")
plt.xlabel("Teorik Kantiller")
plt.ylabel("Örneklem Kantilleri")
 # Üstel dağılım, sadece sağa çarpık (right-skewed) olduğu için, Q-Q grafiğinde S-şekli yerine, 
 # genellikle tek yöne doğru eğri bir yay (bir bumerangı andıran bir eğrilik) oluşturur.

 # S-şekli, genellikle dağılımın hem basıklık hem de çarpıklık açısından Normal dağılımdan saptığı 
 # veya kuyruklarının düzensiz olduğu durumlarda daha belirgin olarak görülür.

plt.tight_layout()
plt.show()
````

### bu grafiklerdeki aykırı değerleri "boxplot" çizerekte görebiliriz. (hoca burada bi durdu)
* Q-Q grafiği size verinin genel olarak normal dağılıp dağılmadığını söylerken, Boxplot size dağılımı bozan aykırı değerlerin (outliers) nerede olduğunu ve kaç tane olduğunu doğrudan gösterir.


Görseldeki iki Q-Q grafiği şu şekilde yorumlanır:

1. **Sol Grafik (Normal Dağılan Veri):**  
   Noktalar neredeyse tam olarak kırmızı referans çizgisi üzerindedir. Bu, verinin normal dağılıma **uyduğunu** gösterir.

2. **Sağ Grafik (Çarpık Dağılan Veri – Üstel Dağılım):**  
   Noktalar referans çizgisinden belirgin şekilde sapar ve **yukarıya doğru kıvrılan bir yay** şekli alır. Bu, verinin **sağa çarpık (right-skewed)** olduğunu ve normal dağılıma **uymadığını** gösterir.

**Özet:**  
Q-Q grafiği, verinin normal dağılıma uyup uymadığını hızlıca değerlendirmek için etkili bir görsel araçtır. Doğrusal çizgi → normallik; eğri çizgi → normallikten sapma.

# verileri normal değılıma uydurma yöntemlerini bil: 
### 1. Parametrik Olmayan (Non-parametrik) Testler

Bu testler, verilerin belirli bir dağılıma (örneğin normal dağılım) uyduğunu varsaymazlar, bu yüzden "dağılımsız" (distribution-free) testler olarak da adlandırılırlar. Parametrik testlerin varsayımları karşılanmadığında güçlü bir alternatiftirler.

**Avantajları:**
*   Normal dağılım varsayımı gerektirmezler.
*   Çarpık, aykırı değerler içeren veya küçük örneklemli verilerle iyi çalışırlar.
*   Nominal ve ordinal (sıralı) veri türleri için de kullanılabilirler.

**Dezavantajları:**
*   Veriler normal dağıldığında, parametrik testlere göre genellikle daha az istatistiksel güce sahiptirler.
*   Sonuçlarının yorumlanması bazen daha zor olabilir.

**Yaygın Parametrik Olmayan Testler ve Parametrik Karşılıkları:**

| Parametrik Test | Parametrik Olmayan Alternatifi | Amaç |
| :--- | :--- | :--- |
| Bağımsız Örneklem t-Testi | **Mann-Whitney U Testi** | İki bağımsız grubun medyanlarını karşılaştırmak. |
| Bağımlı Örneklem t-Testi | **Wilcoxon İşaretli Sıralar Testi** | İki bağımlı grubun (örneğin, ön-test ve son-test) medyanlarını karşılaştırmak. |
| Tek Yönlü ANOVA | **Kruskal-Wallis Testi** | İkiden fazla bağımsız grubun medyanlarını karşılaştırmak. |
| Pearson Korelasyon | **Spearman Sıra Korelasyonu** | İki değişken arasındaki monotonik ilişkiyi ölçmek. |

### 2. Veri Dönüşümleri (Data Transformations)

Bu yöntem, verilerin dağılımını normal dağılıma yaklaştırmak için matematiksel fonksiyonlar uygulamayı içerir. Dönüşümden sonra, standart parametrik testler kullanılabilir.

**Yaygın Dönüşüm Yöntemleri:**

*   **Logaritmik Dönüşüm:** Özellikle sağa çarpık (pozitif çarpık) verilerde etkilidir. Veri dağılımını daha simetrik hale getirebilir ve aykırı değerlerin etkisini azaltabilir.
*   **Karekök Dönüşümü:** Logaritmik dönüşüme göre daha ılımlı bir etkiye sahiptir ve sıfır değerlerine uygulanabilir olması bir avantajdır.

# Varyansların Homojenliği (Eşitliği) Kontrolü - Levene Testi

###hocanın açıklaması 
## Varyansların Homojenliği (Eşitliği) Kontrolü - Levene Testi</br>

**Neden Gerekli?**</br>
T-Testi ve ANOVA gibi Parametrik testler, karşılaştırılan grupların popülasyon varyanslarının istatistiksel olarak eşit olduğunu varsayar. Eğer varyanslar çok farklı ise (heterojen), testin sonuçları (özellikle p-değeri) güvenilirliğini kaybedebilir.</br>

**Kullanım Amacı**</br>
İki veya daha fazla grubun varyanslarının (veya standart sapmalarının) istatistiksel olarak eşit olup olmadığını test etmek için kullanılır. </br>

**Hipotez:**</br>
<ul>
    <li>H0 (Null Hipotezi): Grupların popülasyon varyansları eşittir (Homojen). (p > 0.05 ise kabul edilir)</li>
    <li>H1 (Alternatif Hipotez): En az bir grubun popülasyon varyansı diğerlerinden farklıdır (Heterojen). (p < 0.05 ise H0 reddedilir)</li>
</ul>



````python
# hocanın örneği 
import numpy as np
from scipy import stats
# --- Gerçek Dünya Verisi Senaryosu: İki farklı yatırım stratejisinin getirileri ---
np.random.seed(20)

# Strateji 1: Düşük Risk (Küçük Varyans)
getiri_risk_dusuk = np.random.normal(loc=10, scale=3, size=40) 

# Strateji 2: Yüksek Risk (Büyük Varyans)
getiri_risk_yuksek = np.random.normal(loc=12, scale=10, size=40)

### B. Levene Testi Uygulaması

# center='median' (varsayılan) kullanmak, verinin Normal dağılmadığı durumlarda 
# daha sağlam (robust) sonuçlar verir (Brown-Forsythe Testi'ne yakındır).
levene_stat, levene_p = stats.levene(getiri_risk_dusuk, getiri_risk_yuksek, center='median') # genellikle "medyan kulklanılır"

print(f"Grup 1 Varyansı: {getiri_risk_dusuk.var():.2f}")
print(f"Grup 2 Varyansı: {getiri_risk_yuksek.var():.2f}")
print("-" * 50)
print(f"Levene İstatistiği: {levene_stat:.3f}")
print(f"P Değeri: {levene_p:.5f}")

# Hipotez Karar Mekanizması (α = 0.05)
if levene_p < 0.05:
    print("\nSonuç: P değeri < 0.05. H0 Reddedildi. Grupların varyansları ANLAMLI OLARAK FARKLI (Heterojen). ❌")
    print("Not: Bu durumda T-Testi yapılırken 'equal_var=False' (Welch's T-Test) kullanılmalıdır.")
else:
    print("\nSonuç: P değeri ≥ 0.05. H0 Kabul Edildi. Grupların varyansları EŞİT (Homojen). ✅")
    print("Not: Bu durumda T-Testi yapılırken 'equal_var=True' kullanılabilir.")
````

# dun testi nasıl yapılır? "sınavda bundan sorumluyuz. hoca şimdi söyledi :OO:O:O:OO:O:OO:O:O:O"

**aşağıya bu konuyla alakalı iki tane yapay zeka çıktısı koyuyorum**

### Kruskal-Wallis Testi ve Neden Dunn Testi'ne İhtiyaç Duyarız?

*   **Kruskal-Wallis Testi**, ikiden fazla bağımsız grubun medyanlarının birbirinden farklı olup olmadığını anlamak için kullanılır. Kendisi, Tek Yönlü ANOVA'nın non-parametrik alternatifidir.
*   Kruskal-Wallis testinin sonucu anlamlı çıktığında (yani p ≤ 0.05), bize sadece şunu söyler: "Bu gruplardan en az biri diğerlerinden farklıdır."
*   Ancak bu test, **hangi grubun hangisinden farklı olduğunu** bize söylemez. Örneğin, A, B ve C gruplarını karşılaştırdıysak, fark A ile C arasında mı? B ile C arasında mı? Yoksa her üçü de birbirinden mi farklı?

İşte bu noktada **Dunn Testi** devreye girer.

### Dunn Testi Nedir?

**Dunn Testi**, Kruskal-Wallis Testi sonucunda gruplar arasında anlamlı bir fark bulunduğunda, bu farkın **hangi gruplar arasında** olduğunu tespit etmek için kullanılan bir non-parametrik post-hoc (çoklu karşılaştırma) testidir.

Kısacası, Kruskal-Wallis Testi'nin "fark var" dediği yerde, Dunn Testi "fark şunlar arasında" diyerek detayı ortaya çıkarır.

**Temel Özellikleri:**

1.  **Non-Parametriktir:** Normal dağılım veya varyans homojenliği gibi varsayımlara ihtiyaç duymaz.
2.  **Çoklu Karşılaştırma Düzeltmesi:** Birden fazla ikili karşılaştırma yapıldığı için (A vs B, A vs C, B vs C gibi), Tip I hata (gerçekte fark yokken var bulma) oranını şişirme riski vardır. Dunn Testi, bu hatayı kontrol altında tutmak için kendi içinde bir **p-değeri düzeltme yöntemi** (genellikle Bonferroni veya Benjamini-Hochberg) kullanır. Bu, testin en önemli avantajlarından biridir.
3.  **Sıra Ortalamalarını Kullanır:** Karşılaştırmaları, Kruskal-Wallis testinde de kullanılan "sıra ortalamaları" (mean ranks) üzerinden yapar.

### Dunn Testi Nasıl Yapılır ve Yorumlanır? (Adım Adım)

Dunn Testi'ni uygulamak ve yorumlamak için izlenecek yol şöyledir:

**Adım 1: Ana Testi Uygulayın (Kruskal-Wallis)**

*   Öncelikle, ikiden fazla bağımsız grubunuzun (örneğin, 3 farklı ilacın ağrı kesme skorları) medyanlarının birbirinden farklı olup olmadığını test etmek için **Kruskal-Wallis Testi**'ni yapın.

**Adım 2: Kruskal-Wallis Sonucunu Yorumlayın**

*   Eğer Kruskal-Wallis testinin p-değeri > 0.05 ise, gruplar arasında anlamlı bir fark yoktur. Bu durumda **Dunn Testi'ne gerek kalmaz** ve analiz burada biter.
*   Eğer p-değeri ≤ 0.05 ise, gruplar arasında anlamlı bir fark vardır. Şimdi bu farkın kaynağını bulmak için **Adım 3'e geçin**.

**Adım 3: Dunn Testi'ni Uygulayın**

*   SPSS, R veya Python gibi istatistiksel yazılımlar, Kruskal-Wallis Testi'nin bir seçeneği olarak veya ayrı bir komutla Dunn Testi'ni kolayca uygulamanızı sağlar.
*   Test, olası tüm grup çiftlerini (A-B, A-C, B-C vb.) karşılaştırır ve her bir ikili karşılaştırma için ayrı bir **düzeltilmiş p-değeri (Adjusted p-value)** üretir.

**Adım 4: Dunn Testi Sonuçlarını Yorumlayın**

*   Dunn Testi'nin çıktısı genellikle bir tablo şeklinde olur. Bu tabloda her bir grup çifti için bir p-değeri bulunur.
*   Her bir ikili karşılaştırma için p-değerini yine 0.05 anlamlılık düzeyi ile karşılaştırın:
    *   **Eğer düzeltilmiş p ≤ 0.05 ise:** Bu iki grup arasında istatistiksel olarak anlamlı bir fark vardır.
    *   **Eğer düzeltilmiş p > 0.05 ise:** Bu iki grup arasında istatistiksel olarak anlamlı bir fark yoktur.

#### Örnek Senaryo:

Diyelim ki A, B ve C ilaçlarının baş ağrısını geçirme sürelerini karşılaştırdık ve Kruskal-Wallis Testi p=0.02 çıktı. Bu, "fark var" demektir. Şimdi Dunn Testi'ni uyguladık ve şu sonuçları aldık:

| Grup Karşılaştırması | Düzeltilmiş p-değeri | Yorum |
| :--- | :--- | :--- |
| A - B | 0.89 | Fark yok (p > 0.05) |
| A - C | **0.01** | **Anlamlı fark var (p ≤ 0.05)** |
| B - C | **0.04** | **Anlamlı fark var (p ≤ 0.05)** |

**Nihai Sonuç:** A, B ve C ilaçları arasında genel bir fark vardır. Dunn Testi sonuçlarına göre bu fark, hem A ilacı ile C ilacı arasında, hem de B ilacı ile C ilacı arasında anlamlıdır. Ancak A ve B ilaçları arasında anlamlı bir fark bulunmamaktadır.

### Dunn Testi vs. Diğer Post-Hoc Testler

*   **Mann-Whitney U Testi'ni Tekrarlamak Neden Yanlış?**
    Bazı araştırmacılar, Kruskal-Wallis sonrası her grup çiftine ayrı ayrı Mann-Whitney U testi uygular. Bu, **Tip I hata oranını ciddi şekilde şişirdiği için yanlıştır**. Dunn Testi, p-değeri düzeltmesi yaptığı için bu hatayı önler ve bu yüzden doğru yöntemdir.

*   **Dunn Testi vs. Conover-Iman Testi**
    Conover-Iman da bir başka non-parametrik post-hoc testidir. Genellikle Dunn Testi'nden biraz daha güçlü (farkı bulma olasılığı daha yüksek) kabul edilir, ancak her iki test de yaygın olarak kullanılır.

Özetle, **Dunn Testi**, Kruskal-Wallis'in bulduğu genel farkın "adresini" gösteren, vazgeçilmez bir non-parametrik analiz aracıdır.

---

**Dunn Testi** (İngilizce: *Dunn’s post-hoc test*), **non-parametrik** bir çoklu karşılaştırma (post-hoc) testidir. Temel amacı, **Kruskal-Wallis H testi** sonucunda gruplar arasında anlamlı bir fark bulunduğu tespit edildikten sonra, **hangi gruplar arasında fark olduğunu** belirlemektir.

---

## 🔍 1. **Neden Dunn Testi Gerekir?**

- **Kruskal-Wallis testi**:  
  - 3 ya da daha fazla **bağımsız grubun** medyanlarının eşit olup olmadığını test eder.  
  - Sonuç **genel bir fark** olup olmadığını söyler, **ama hangi gruplar arasında fark olduğunu söylemez**.

- **Dunn testi**, bu genel farkın kaynağını **ikili grup karşılaştırmalarıyla** inceler.

> 📌 **Özetle**:  
> Kruskal-Wallis → “En az bir grup farklı.”  
> Dunn testi → “1. grup ile 3. grup farklı, ama 1-2 ve 2-3 aynı.”

---

## 🧠 2. **Dunn Testinin Mantığı**

1. Tüm veriler **birleştirilip sıralanır** (rank yapılır).
2. Her grup için **ortalama rank** hesaplanır.
3. Her grup çifti arasında **rank farkı** temel alınarak bir **Z istatistiği** hesaplanır.
4. Bu Z istatistiğine göre **p-değeri** bulunur.
5. **Çoklu karşılaştırma düzeltmesi** yapılır (örneğin Bonferroni, Holm).

> ✅ Veri normal dağılmadığı için **ortalamalar değil**, **rank’lar** kullanılır.  
> ✅ Parametrik olmayan verilerde güvenilir sonuç verir.

---

## 📐 3. **Formül (Basitleştirilmiş)**

Grup \(i\) ve \(j\) karşılaştırıldığında:

\[
Z = \frac{\bar{R}_i - \bar{R}_j}{\sqrt{ \frac{N(N+1)}{12} \left( \frac{1}{n_i} + \frac{1}{n_j} \right) }}
\]

- \(\bar{R}_i, \bar{R}_j\): i. ve j. grubun ortalama rankları  
- \(n_i, n_j\): grup büyüklükleri  
- \(N\): toplam gözlem sayısı

> Bu Z skoru, standart normal dağılıma göre p-değeri hesaplamak için kullanılır.

---

## 🛠️ 4. **Dunn Testi Nasıl Uygulanır? (Python Örneği)**

### 🔸 Gerekli kütüphaneler:
```bash
pip install scikit-posthocs pandas
```

### 🔸 Kod:
```python
import numpy as np
import scipy.stats as stats
import scikit_posthocs as sp
import pandas as pd

# Örnek veri: 3 farklı ilaç grubundan elde edilen ağrı skorları (non-normal)
np.random.seed(42)
grup_A = np.random.exponential(2, 30)  # Sağa çarpık
grup_B = np.random.exponential(3, 30)
grup_C = np.random.exponential(4, 30)

# Kruskal-Wallis testi (önce genel fark var mı?)
H, p = stats.kruskal(grup_A, grup_B, grup_C)
print(f"Kruskal-Wallis p-değeri: {p:.4f}")

if p < 0.05:
    print("→ Gruplar arasında anlamlı fark var. Dunn testi uygulanıyor...\n")
    
    # Veriyi DataFrame formatına getir (scikit-posthocs için)
    data = np.concatenate([grup_A, grup_B, grup_C])
    groups = ['A']*30 + ['B']*30 + ['C']*30
    df = pd.DataFrame({'skor': data, 'grup': groups})
    
    # Dunn testi (Bonferroni düzeltmeli)
    dunn_result = sp.posthoc_dunn(df, val_col='skor', group_col='grup', p_adjust='bonferroni')
    print("Dunn Testi Sonuçları (p-değerleri):")
    print(dunn_result)
else:
    print("→ Gruplar arasında fark yok. Dunn testi gerekmez.")
```

### 🔸 Çıktı Yorumu:
- Tablodaki her hücre, iki grup arasındaki **düzeltmeli p-değeri**dir.
- Örneğin, `A-B` hücresi = 0.08 → fark **anlamlı değil** (α = 0.05).
- `A-C` hücresi = 0.001 → fark **anlamlı**.

---

## ⚠️ 5. **Önemli Notlar**

| Konu | Açıklama |
|------|--------|
| **Veri varsayımları** | Bağımsız gruplar, sıralanabilir (ordinal ya da sürekli) veri |
| **Normallik gerekli mi?** | ❌ Hayır — zaten non-parametrik bir test |
| **Eşit varyans gerekli mi?** | ❌ Hayır — Dunn testi varyans homojenliğine gerekmez |
| **Çoklu test düzeltmesi** | ✅ **Zorunlu** — aksi halde Type I hata oranı artar. En yaygın: `bonferroni`, `holm`, `fdr` |

---

## 🆚 6. **Dunn Testi vs. Diğer Post-Hoc Testleri**

| Test | Parametrik mi? | Öncül Test | Kullanım |
|------|----------------|-----------|--------|
| **Tukey HSD** | Evet | ANOVA | Normal veri, eşit varyans |
| **Games-Howell** | Evet (kısıtlı) | ANOVA | Normal veri, **eşit olmayan varyans** |
| **Dunn Testi** | ❌ Hayır | Kruskal-Wallis | **Non-normal veri** |
| **Conover Testi** | ❌ Hayır | Kruskal-Wallis | Dunn’a alternatif; bazen daha güçlü |

---

- **Dunn testi**, Kruskal-Wallis sonrası **non-parametrik çoklu karşılaştırma** yapar.
- **Rank farklarına** dayanır.
- **Çoklu karşılaştırmada p-değeri düzeltmesi** mutlaka yapılmalıdır.
- Python’da `scikit-posthocs.posthoc_dunn()` ile kolayca uygulanır.
